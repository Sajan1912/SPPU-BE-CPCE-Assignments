

Question 1 :

Design an unequal ISA to resist the tensile force of 225kN as per IS 800.
Use M20 bolts. Assume suitable data if required.



```python
# Design of tension member

# Input values
Tu = float(input("Enter the value of ultimate tensile force (kN): "))
fy = float(input("Enter the value of yield strength of steel (MPa): "))
fu = float(input("Enter the value of ultimate strength of steel (MPa): "))
fub = float(input("Enter the value of ultimate strength of bolt (MPa): "))
Gamma_mo = float(input("Enter the value of partial factor of safety Gamma_mo: "))
Gamma_m1 = float(input("Enter the value of partial factor of safety Gamma_m1: "))
Gamma_mb = float(input("Enter the value of partial factor of safety Gamma_mb: "))

# Gross Area Required
Agreq = (1.1 * Tu * 1000) / fy
print("The gross area required is: ", Agreq, "mm²")

# Selection of section (example ISA 100x65x8 is used here)
Ag = 1257  # Example gross area (mm²)
Lcl = float(input("Enter the length of connected leg (mm): "))
Lol = float(input("Enter the length of outstand leg (mm): "))
t = float(input("Enter the value of thickness (mm): "))

# Design of connections
d = float(input("Enter the value of bolt diameter (mm): "))
do = d + 2
print("The diameter of bolt hole is: ", do, "mm")

# IS code minimum pitch distance
pmin = 2.5 * d
print("The minimum pitch distance is: ", pmin, "mm")

# Edge distance as per IS 800
e = 1.5 * do
print("The edge distance is: ", e, "mm")

# Number of shear planes
nn = float(input("Enter the number of shear planes with threads intercepting the shear plane: "))
ns = float(input("Enter the number of shear planes without threads: "))

# Bolt area
Anb = 0.78 * (3.1416 * d**2) / 4
Asb = (3.1416 * d**2) / 4
print("Threaded area of bolt: ", Anb, "mm²")
print("Plane shank area of bolt: ", Asb, "mm²")

# Shear strength of bolt
Vdsb = (fub / (1.732 * Gamma_mb)) * (nn * Anb + ns * Asb) * 10**-3
print("Shear strength of bolt (Vdsb): ", Vdsb, "kN")

# Calculation of kb
kb1 = e / (3 * do)
kb2 = (pmin / (3 * do)) - 0.25
kb3 = fub / fu
kb4 = 1
kb = min(kb1, kb2, kb3, kb4)
print("Kb: ", kb)

# Bearing strength of bolt
Vdpb = (2.5 * kb * d * t * fu * 10**-3) / Gamma_mb
print("Bearing strength of bolt (Vdpb): ", Vdpb, "kN")

# Design strength of the bolt
Vd = min(Vdsb, Vdpb)
print("Design strength of bolt (Vd): ", Vd, "kN")

# Number of bolts required
N = Tu / Vd
print("Number of bolts required: ", N)

# Check for strength
# 1. Yielding of Gross Section
Tdg = (Ag * fy * 10**-3) / Gamma_mo
print("Tensile strength due to yielding of gross section (Tdg): ", Tdg, "kN")

# 2. Rupture of net section
Anc = (Lcl - (t / 2) - do) * t
Ago = (Lol - (t / 2)) * t
print("Net area of connected leg (Anc): ", Anc, "mm²")
print("Gross area of outstand leg (Ago): ", Ago, "mm²")

# Effective length
Le = (N - 1) * pmin
print("Effective length (Le): ", Le, "mm")

# Beta calculation
bs = 0.6 * Lcl + Lol
Beta = 1.4 * (0.076 * (fy / fu) * (bs / Le))
print("Beta: ", Beta)

# Check safety conditions for Beta
if Beta > 1.4:
    print("Not Safe")
else:
    print("Safe")

if Beta < 0.7:
    print("Not Safe")
else:
    print("Safe")

# Net tension strength (Tdn)
Tdn = ((0.9 * fu * Anc) / Gamma_m1) + (Beta * Ago * fy / Gamma_mo)
print("Net tension strength (Tdn): ", Tdn, "kN")

# 3. Block shear strength
Avg = (pmin * (N - 1) + e) * t
Avn = ((pmin * (N - 1) + e) - (N - 1) * do + (8.5 * do)) * t
Atg = 0.6 * Lcl * t
Atn = Atg - 0.5 * do
print("Avg: ", Avg, "mm²")
print("Avn: ", Avn, "mm²")
print("Atg: ", Atg, "mm²")
print("Atn: ", Atn, "mm²")

# Block shear strength calculations
Tb1 = ((Avg * fy) / (1.732 * Gamma_mo)) + (0.9 * fu * Atn) / Gamma_m1
Tb2 = (0.9 * Avn * fu / (1.732 * Gamma_m1)) + (Atg * fy) / Gamma_mo
Tb = min(Tb1, Tb2)
print("Block shear strength (Tb): ", Tb, "kN")

# Final tensile strength
Td = min(Tdg, Tdn, Tb)
print("Final design strength (Td): ", Td, "kN")

# Check if the section is safe
if Td > Tu:
    print("SAFE")
else:
    print("Revise the Section")
```
