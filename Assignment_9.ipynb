

Question 1 :

Determine the BOD on the 7th day at 25°C if the BOD on the 3rd day at 20°C is found to be 50 mg/L. Take the decay coefficient as 0.23.



```python
# To find BOD at 7th day 25°C
import math

# Decay coefficient at 20°C
K1 = float(input("Decay Coefficient at 20°C: "))
T1 = float(input("Temperature of 3rd day BOD (in °C): "))
T2 = float(input("Temperature of 7th day BOD (in °C): "))

# To calculate decay coefficient at 25°C (K2)
K2 = K1 * (1.047 ** (T2 - T1))
print("The value of K2 (Decay Coefficient at 25°C) is:", K2)

# To find the ultimate BOD at 3rd day 20°C (B1)
B1 = float(input("BOD at 3rd day at 20°C (in mg/L): "))
t = 3  # Time for B1, which is 3 days
# Calculate the fraction of BOD exerted in 3 days
E = 1 - math.exp(-K1 * t)
print("The fraction of BOD exerted at 3rd day (E):", E)

# Ultimate BOD (L0) using BOD at 3rd day
L0 = B1 / E
print("The value of Ultimate BOD (L0) is:", L0)

# To find BOD at 7th day 25°C
t1 = 7  # Time for BOD at 7 days
# Calculate the fraction of BOD exerted in 7 days at 25°C
E1 = 1 - math.exp(-K2 * t1)
print("The fraction of BOD exerted at 7th day (E1):", E1)

# Calculate BOD at 7th day at 25°C (B2)
B2 = L0 * E1
print("The BOD at 7th day at 25°C (B2) is:", B2, "mg/L")
```

The sludge from the aeration tank of the Activated Sludge Process (ASP) has a solid content by weight of 2%. The sludge is put in a sludge thickener where the sludge volume is reduced to half. Assume that the amount of solids in the supernatant from the thickener is negligible. The specific gravity of sludge solids is 2.2, and the density of water is 1000 kg/m³. What is the density of the sludge removed from the aeration tank? (Hint: Assume the initial mass of sludge is 100 kg).


```python
# Determination of density of sludge removed from aeration tank

M = float(input("Enter the value of initial mass (kg): "))  # Initial mass of sludge
S = float(input("Enter the solid content of sludge in percentage: "))  # Solid content in percentage
Gs = float(input("Enter the specific gravity of sludge solid: "))  # Specific gravity of sludge solids
Rho_W = float(input("Enter the density of water (kg/m³): "))  # Density of water

# Calculating mass of solids in the sludge
Ws = (S / 100) * M  # Solid content in sludge (kg)
m = M - Ws  # Mass of water in the sludge (kg)

print("The mass of water in sludge: ", m, "kg")
print("The solid content in sludge: ", Ws, "kg")

# Calculating the volume of water in the sludge
Vw = m / Rho_W  # Volume of water in sludge (m³)
print("The volume of water in sludge: ", Vw, "m³")

# Calculating the density of the solid content in sludge
Rho_S = Gs * Rho_W  # Density of solid content in sludge (kg/m³)
print("The density of solid content in sludge: ", Rho_S, "kg/m³")

# Calculating the volume of solid content in sludge
Vs = Ws / Rho_S  # Volume of solid content in sludge (m³)
print("The volume of solid content in sludge: ", Vs, "m³")

# Total volume of the sludge
Vt = Vw + Vs  # Total volume of the sludge (m³)
print("The total volume of sludge: ", Vt, "m³")

# Density of the sludge removed from aeration tank
Rho_SL = M / Vt  # Density of sludge (kg/m³)
print("The density of sludge removed from aeration tank: ", Rho_SL, "kg/m³")
```
