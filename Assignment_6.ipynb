

Question 1 :

A concentrated load of 2500 kN is applied at the ground surface. Determine the vertical stress at a point up to 6 m depth and at a horizontal distance of 5 m from the axis of the load. Plot the graph by using Excel. Use Boussinesq's Theory.



```python
# Stress When depth is constant
Q = float(input("Enter the value of Load in kN: "))
N = int(input("Number of data values of radial distance: "))
pi = 3.14159265359
Z = float(input("Depth in meters: "))

r = []
for i in range(1, N + 1):
    print("Enter radial distance in meters:")
    Value_r = float(input())
    r.append(Value_r)

    # Calculate stress using Boussinesq's formula
    Stress = ((3 * Q) / (2 * pi * Z * Z)) * (((1 / (1 + ((Value_r / Z) ** 2))) ** 2.5))

    print("Stress: {:.4f} kN/m^2".format(Stress))
```
